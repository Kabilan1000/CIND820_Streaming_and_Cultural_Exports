# Abstract

This project examines how Netflix, Amazon Prime Video, and Disney+ act as cultural intermediaries by shaping the availability and visibility of global media. The analysis combines exploratory data methods with metadata-based prediction to study catalogue composition, representation patterns, and the relationship between content characteristics and popularity, all defined using IMDb ratings.

The study uses platform, type, release year, and the top twenty genres as predictors. Logistic regression and random forest models are evaluated against a majority-class baseline with stratified cross-validation to test effectiveness, efficiency, and stability. The results show clear differences in catalogue structures across platforms, alongside modest but consistent predictive signal in the selected metadata features. Furthermore, platform-specific patterns also emerge, indicating that certain content characteristics relate to popularity differently depending on the streaming service.

Since country-of-origin metadata is incomplete and inconsistent, internationally oriented genres are used as a structured proxy to approximate cultural diversity. All in all, the findings show how catalogue construction and metadata patterns influence cultural representation across major streaming platforms.

# Methodology and Study Design

## Data Sources

This study combines three publicly available Kaggle datasets—Netflix, Amazon Prime Video, and Disney+, alongside metadata obtained from the official IMDb datasets. The Kaggle data provides information on title names, genres, runtime, release years, and content type (movie or TV show). IMDb supplies additional fields including title basics, ratings, and alternate titles (AKAs), which support the popularity label and limited country-of-origin extraction.

OMDb was considered initially, but its rate limits and inconsistent coverage made it unsuitable for large-scale merging. IMDb’s dumps allowed complete offline processing, and all merges were performed programmatically.

## Data Cleaning and Integration

The cleaning process focused on standardizing the three platform datasets before aligning them with IMDb:

>- Titles were normalized by lowercasing and removing excess punctuation.
>- Duplicate entries were removed, with attention to titles appearing on multiple platforms.
>- Content types were harmonized into two categories: *Movie* and *TV Show*.
>- Release years were cleaned to remove invalid or missing values.
>- Genres were parsed into list form and later expanded into indicator columns.
>- Country-of-origin data from IMDb AKAs was retained only for descriptive plots because of inconsistency and missingness.

To improve efficiency, the merged streaming–IMDb dataset was cached. This avoided recomputing heavy joins during consecutive modeling.

## Feature Engineering

The predictive features include:

- **Platform**: One-Hot Encoded (Netflix, Amazon Prime Video, Disney+)
- **Type**: Movie vs TV Show
- **Release Year**: Standardized as a numerical feature
- **Top Twenty Genres**: Converted into binary indicator variables

Only the most frequent genres were kept as predictors to reduce sparsity and noise from the rarely occurring categories. Country metadata was excluded from modeling due to low quality and limited coverage.
>The target variable, *popular*, was defined using the following IMDb ratings:
>>- `popular = 1` if rating ≥ 7.0  
>>- `popular = 0` otherwise

The rating itself was used strictly as the label and never included as a predictor, avoiding any form of leakage.

## Modeling Pipeline

Two models were selected:

- **Logistic Regression** as an interpretable baseline
- **Random Forest** as a non-linear benchmark capable of capturing interactions

A **majority-class baseline** served as a reference point. The pipeline used:

>- One-hot encoding for categorical variables  
>- Standardization for numerical features  
>- Stratified train/test split (70/30)  
>- Five-fold **stratified cross-validation** for stability  

Logistic regression coefficients were further evaluated using bootstrap resampling to examine coefficient stability and identify signals that consistently remained positive or negative. For Random Forest, feature importances were calculated to understand which metadata fields carried the strongest predictive signal.

## Evaluation Metrics

Model performance was measured using:

>- Accuracy  
>- Precision  
>- Recall  
>- F1 score  
>- Confusion matrices  
>- Cross-validation mean and standard deviation  

These metrics capture both the overall correctness of predictions and the model’s ability to identify titles labeled as popular. Efficiency was assessed using model fit times and the impact of the cached IMDb merge.

Together, these methods form a transparent and reproducible framework that integrates exploratory analysis with metadata-driven prediction while maintaining attention to stability and methodological reliability.

# Model Evaluation

## Effectiveness

The effectiveness of each model was evaluated against a majority-class baseline. Since the dataset is imbalanced, with most titles not meeting the “popular” threshold, the baseline accuracy is approximately 69%. This baseline, however, produces an F1 score of 0 for the positive class because it never predicts popularity.

Both logistic regression and random forest outperform this baseline. Logistic regression achieves an accuracy in the low 70% range, with measurable improvements in precision and recall for the popular class. Random forest provides a similar level of uplift. Performing slightly better on recall and overall F1 score. While the absolute gains are modest, they are consistent across splits and indicate that platform, type, release year, and the most frequent genres do carry meaningful predictive signal.

Confusion matrices for both models show that the improvements come primarily from a better ability to identify popular titles, rather than simply reinforcing the class imbalance. As such the interpretation that metadata alone offers limited but genuine predictive structure is supported.

## Efficiency

The models' efficiency was evaluated with two considerations in mind: Each model's runtime and the overhead of preparing the merged dataset for evaluation. The cleaning and merge steps involving IMDb data were the most computationally expensive part of the workflow. To address this requirement, the final merged dataset was cached.  THis helped buy eliminating the need to repeatedly recompute the joins and significantly sped up repeated modeling experiments.

As expected Logistic Regression had the fastest training time due to its linear nature and small feature space. Random Forest on the other hand required more computation but remained well within practical limits for this dataset. Overall, the efficiency of the workflow was enough to run interactively and supports rapid iteration once the first initial merge is cached.

## Stability

Stability was tested using stratified five-fold cross-validation. Logistic regression produced stable performance across mutliple folds, with small variance in both accuracy and F1 score. Random forest showed slightly higher variance, which is expected for ensemble models.  All in all the results remained consistent to confirm that the signal in the metadata generalizes beyond a single train/test split.

Furthermore there was an additional check provided by the bootstrap resampling of logistic regression coefficients. Several coefficients, including those connected to platform, content type, and a subset of genres, assisted in displaying tight confidence intervals. This resulted in evidence showing that their direction and magnitude are not artifacts of a single sample. Random forest feature importances supported these results as well, ranking release year, content type, and certain other genres as the most influential predictors overall.

The cross-validation results, alongside bootstrapped coefficients, and importance rankings indicate that the observed patterns are stable and not driven by noise or overfitting. While the overall predictive power remains somewhat moderate, the consistency across models and validation methods confirms that the metadata features do contribute to the meaningful structure of the popularity label.

# Limitations and Ethical Considerations

## Data Limitations

The dataset combines three separate Kaggle catalogues alongside metadata extracted from IMDb, which results in several limitations that follow from this integration. Firstly, the country-of-origin metadata is vastly incomplete, inconsistent across titles, and difficult to standardize. Thus the field was made unreliable for modeling by multi-country productions and missing values, restricting it to only descriptive use.

Second, IMDb ratings are purely user-generated and do not represent actual viewership. They reflect a group of users who are self-selected and can thus introduce demographic or regional biases. The fact that only titles with available IMDb ratings can be included further constrains the modeling subset, which may end up skewing the sampling toward more visible or better-known works.

Finally, the genre labels across platforms can vary in specificity. Certain platforms may apply a series of broad genre categories, while others will list multiple overlapping tags, and although the top twenty genres provided reasonable coverage, the sparsity problem remains an issue for the less common genres.

## Methodological Limitations

The models rely entirely on metadata, and so without the textual descriptions, casting information, or overall recommendation data, the models can only capture high-level structural signals. This results in the predictive power remaining somewhat moderate and is not intended to represent deeper audience-based dynamics and demographics.

Furthermore, the results are associative rather than causal. Logistic regression coefficients and random forest importances show which features correlate with popularity but they cannot explain why these those relationships exist. As well, cross-validation and bootstrap checks can help ensure stability but do not eliminate the inherent limits of a correlational analysis.

The 70/30 split and five-fold stratified cross-validation, while sufficient for this dataset, a more advanced time-based validation or platform-specific splits would be able tonprovide additional insights.

## Ethical Considerations

Using IMDb ratings as a popularity proxy introduces a series of potential cultural and demographic biases, as there are often rating patterns can vary widely across regions and user communities. These biases may influence which titles are labeled as “popular,” especially when looking at international content.

The project uses only publicly available metadata and does not involve any personal or identifiable user information, thus reducing privacy concerns. However, the broader implications of individual platform-driven visibility should still be acknowledged. Streaming services serve a gatekeeper role, an as such any analysis of popularity or representatin should consider how a platform's algorithms, licensing decisions, and catalogue design can affect overall cultural exposure.

All in all, the study is restricted by the quality and structure of the available data, but still remains to be methodologically transparent and ethically responsible within those the stated boundaries.
