### **Project Title:**  
**Streaming Platforms as Cultural Intermediaries: An Exploratory Analysis of Netflix, Amazon Prime Video, and Disney+**


# Abstract

This project examines how Netflix, Amazon Prime Video, and Disney+ act as cultural intermediaries by shaping the availability and visibility of global media. The analysis combines exploratory data methods with metadata-based prediction to study catalogue composition, representation patterns, and the relationship between content characteristics and popularity, all defined using IMDb ratings.

The study uses platform, type, release year, and the top twenty genres as predictors. Logistic regression and random forest models are evaluated against a majority-class baseline with stratified cross-validation to test effectiveness, efficiency, and stability. The results show clear differences in catalogue structures across platforms, alongside modest but consistent predictive signal in the selected metadata features. Furthermore, platform-specific patterns also emerge, indicating that certain content characteristics relate to popularity differently depending on the streaming service.

Since country-of-origin metadata is incomplete and inconsistent, internationally oriented genres are used as a structured proxy to approximate cultural diversity. All in all, the findings show how catalogue construction and metadata patterns influence cultural representation across major streaming platforms.

# 3. Introduction

Streaming platforms have become the primary gateway that global audiences use to access film and television content. Services such as Netflix, Amazon Prime Video, and Disney+ play a critical role in determining what cultural material is made visible, how it is organized, and which types of content are amplified or sidelined. Their catalogues function as curated ecosystems that shape both cultural exposure and audience expectations. As a result, understanding how platform catalogues differ by, genre, release period, and overall content mix, is essential for evaluating their influence on global media flows.

Popularity is another key dimension for evaluation. A platform’s catalogue may be diverse, but visibility can be unequally distributed. Titles can differ in how often they are recommended, discovered, and talked about. While true viewership data is not publicly available, IMDb ratings can serve as a widely used proxy for audience reception. Using metadata only, this project tests whether basic platform and content characteristics carry enough signal to predict which titles are more likely to achieve higher popularity via IMDb scores. This offers a transparent way to examine how platform-specific catalogue structures can relate to popularity outcomes.

The third issue involves cultural diversity. Ideally, this would be studied using country-of-origin metadata, but when put into practice these fields were incomplete, inconsistent, and difficult to standardize across different datasets. As such, to address this, the analysis uses internationally oriented genres, such as *International Movies*, *Anime*, and *Foreign Documentaries*, as a structured approximation of cultural breadth across platforms. This avoids forcing unreliable country labels into the modeling process while still enabling a meaningful comparison of global content representation.

Together, these considerations motivate four research questions to guide the study:

1. How do Netflix, Disney+, and Amazon Prime Video differ in their catalogue composition with respect to genre distributions, release-year patterns, and the balance between movies and TV shows?

2. To what extent can platform, type, release year, and genre information predict whether a title is “popular,” using IMDb ratings as the target?

3. Do metadata factors relate to popularity differently across platforms, and which variables show consistent positive or negative associations with predicted popularity?

4. How can internationally oriented genres be used to approximate cultural diversity across platforms, given the unreliability of country-of-origin metadata?

This report investigates these questions through a combination of exploratory analysis and predictive modeling, with an emphasis on transparency, reproducibility, and metadata patterns.
